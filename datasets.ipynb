```python
import zipfile
import os

zip_dir = "/Users/yaminichitikela/Desktop/ISL PROJECT/include_videos"
extract_dir = "/Users/yaminichitikela/Desktop/ISL PROJECT/include_videos"

for fname in sorted(os.listdir(zip_dir)):
    if fname.endswith(".zip"):
        zip_path = os.path.join(zip_dir, fname)
        print(f"Extracting {fname}...")
        with zipfile.ZipFile(zip_path, 'r') as zf:
            zf.extractall(extract_dir)

print("Done!")
```
